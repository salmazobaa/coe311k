```python
import numpy as np

def gaussian_elimination(A, b):
    n = len(b)

    # Forward elimination
    for i in range(n):

        # Pivoting (find biggest value in column)
        max_row = i
        for r in range(i + 1, n):
            if abs(A[r][i]) > abs(A[max_row][i]):
                max_row = r

        # Swap rows in A
        temp = A[i]
        A[i] = A[max_row]
        A[max_row] = temp

        # Swap values in b
        temp_b = b[i]
        b[i] = b[max_row]
        b[max_row] = temp_b

        # Make values below pivot zero
        for j in range(i + 1, n):
            ratio = A[j][i] / A[i][i]

            for k in range(n):
                A[j][k] = A[j][k] - ratio * A[i][k]

            b[j] = b[j] - ratio * b[i]

    # Back substitution
    x = [0] * n

    for i in range(n - 1, -1, -1):
        total = 0
        for j in range(i + 1, n):
            total = total + A[i][j] * x[j]

        x[i] = (b[i] - total) / A[i][i]

    return x


def lu_decomposition(A):
    n = len(A)

    # Make L as identity matrix
    L = np.eye(n)

    # Make a copy of A for U
    U = np.array(A, dtype=float)

    for i in range(n):

        if U[i][i] == 0:
            raise ValueError("Zero pivot encountered")

        for j in range(i + 1, n):
            factor = U[j][i] / U[i][i]
            L[j][i] = factor

            for k in range(i, n):
                U[j][k] = U[j][k] - factor * U[i][k]

    return L, U
```
