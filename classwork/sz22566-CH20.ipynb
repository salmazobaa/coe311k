```python
import numpy as np
import matplotlib.pyplot as plt

# --- 1. Nonlinear Problem Definition ---
def f(t, y):
    return y**2 - 10

def fprime(y):
    # Derivative of f with respect to y: df/dy = 2y
    return 2 * y

# --- 2. Your Provided Step Function ---
def backward_euler_step(f, y, t, h, tol=1e-6, max_iter=50):
    y_next = y  # Initial guess
    for _ in range(max_iter):
        g_val = y_next - y - h * f(t + h, y_next)
        g_prime_val = 1 - h * fprime(y_next)     
        
        if abs(g_prime_val) < tol:
            break
            
        y_new = y_next - g_val / g_prime_val
        
        if abs(y_new - y_next) < tol:
            return y_new
        y_next = y_new
    return y_next

# --- 3. Simulation Loop for Multiple h values ---
y0 = 5.0     # Initial condition
t0 = 0.0
tf = 0.5     # Short time span as y^2 grows quickly
h_list = [0.01, 0.05, 0.1, 0.15, 0.2, 0.25]

plt.figure(figsize=(10, 6))

for h in h_list:
    t_vals = np.arange(t0, tf + h, h)
    y_vals = np.zeros_like(t_vals)
    y_vals[0] = y0
    
    for i in range(1, len(t_values)):
        y_vals[i] = backward_euler_step(f, y_vals[i-1], t_vals[i-1], h)
    
    plt.plot(t_vals, y_vals, 'o-', label=f'h = {h}')

# --- 4. Plotting Results ---
plt.title('Nonlinear ODE (y^2 - 10) using Backward Euler/Newton-Raphson')
plt.xlabel('Time (t)')
plt.ylabel('y(t)')
plt.legend()
plt.grid(True)
plt.show()
```
