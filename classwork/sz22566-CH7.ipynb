```python
import numpy as np

def jacobi(A, b, x0, tol, max_iter=1000):
    n = len(b)
    x = x0.copy()
    errors = []

    for k in range(max_iter):
        x_new = np.zeros(n)

        for i in range(n):
            total = 0
            for j in range(n):
                if j != i:
                    total = total + A[i][j] * x[j]

            x_new[i] = (b[i] - total) / A[i][i]

        error = np.max(np.abs(x_new - x))
        errors.append(error)

        if error < tol:
            return x_new, k + 1, errors

        x = x_new

    return x, max_iter, errors


def gauss_seidel(A, b, x0, tol, max_iter=1000):
    n = len(b)
    x = x0.copy()
    errors = []

    for k in range(max_iter):
        x_old = x.copy()

        for i in range(n):
            sum1 = 0
            for j in range(i):
                sum1 = sum1 + A[i][j] * x[j]

            sum2 = 0
            for j in range(i + 1, n):
                sum2 = sum2 + A[i][j] * x_old[j]

            x[i] = (b[i] - sum1 - sum2) / A[i][i]

        error = np.max(np.abs(x - x_old))
        errors.append(error)

        if error < tol:
            return x, k + 1, errors

    return x, max_iter, errors


A1 = np.array([
[0.582745, 0.48 , 0.10 , 0., 0.],
[0.48 , 1.044129, 0.46 , 0.10 , 0.],
[0.10 , 0.46 , 1.10431 , 0.44 , 0.10],
[0., 0.10, 0.44 , 0.963889, 0.42],
[0., 0., 0.10 , 0.42 , 0.522565]], dtype=float)

b1 = np.array([1.162745, 2.084129, 2.20431 , 1.923889, 1.042565], dtype=float)

x0_1 = np.zeros(5)

jacobi_result1 = jacobi(A1, b1, x0_1, 1e-9)
gs_result1 = gauss_seidel(A1, b1, x0_1, 1e-9)

print("System 1")
print("Jacobi Solution:", jacobi_result1[0])
print("Gauss-Seidel Solution:", gs_result1[0])
print()

A2 = np.array([[1, 1],
               [1, 1.0001]], dtype=float)

b2 = np.array([1, 1], dtype=float)

x0_2 = np.zeros(2)

jacobi_result2 = jacobi(A2, b2, x0_2, 1e-9)
gs_result2 = gauss_seidel(A2, b2, x0_2, 1e-9)

print("System 2")
print("Jacobi Solution:", jacobi_result2[0])
print("Gauss-Seidel Solution:", gs_result2[0])
print()

A3 = np.array([[1, 1],
               [1, 1.0001]], dtype=float)

b3 = np.array([1, 1.0001], dtype=float)

x0_3 = np.zeros(2)

jacobi_result3 = jacobi(A3, b3, x0_3, 1e-9)
gs_result3 = gauss_seidel(A3, b3, x0_3, 1e-9)

print("System 3")
print("Jacobi Solution:", jacobi_result3[0])
print("Gauss-Seidel Solution:", gs_result3[0])
```
