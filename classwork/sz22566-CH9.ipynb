```python

import numpy as np
import matplotlib.pyplot as plt
import random

# Data points

X = np.array([1, 2, 3, 4, 5])

Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])

# Stack the design matrix with a column of ones (for the intercept)

A = np.vstack([X, np.ones(len(X))]).T

# Solve using least squares

m, c = np.linalg.lstsq(A, Y, rcond=None)[0]

print(f"Slope: {m}, Intercept: {c}")

# part a: plotting the data points and the fitted line

plt.figure()
plt.scatter(X, Y, color="blue", label="Data Points")
plt.plot(X, m * X + c, color="red", label="Fitted Line")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression using Least Squares")
plt.legend()
plt.show()


# part b: randomizing vectors then rerun

# randomizing the vectors destroys the relationship betweeb X and Y,
# which destroys the linear relationship and results in a poor fit.

X2 = np.random.randint(1, 101, size = 20)
Y2 = np.random.randint(1, 101, size = 20)

x2 = np.sort(X2)
y2 = np.sort(Y2)

A2 = np.vstack([x2, np.ones(len(x2))]).T
m2, c2 = np.linalg.lstsq(A2, y2, rcond=None)[0]

print(f"Shuffled Slope: {m2}, Intercept: {c2}")

# plotting the data points and the fitted line for shuffled vectors

plt.figure()
plt.scatter(x2, y2, color="blue", label="Data Points")
plt.plot(x2, m2 * x2 + c2, color="red", label="Fitted Line")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Shuffled Vectors Fitted Line")
plt.legend()
plt.show()

# part c: creating my own linear equation

X3 = np.arange(0, 1, 0.01)
perturbation = np.random.uniform(0, 1, len(X3))
Y3 = 2 * X3 + 1 + perturbation

# Stack the design matrix with a column of ones (for the intercept)

A3 = np.vstack([X3, np.ones(len(X3))]).T

# Solve using least squares

m3, c3 = np.linalg.lstsq(A3, Y3, rcond=None)[0]

print(f"Slope: {m3}, Intercept: {c3}")

# residuals
residuals = np.abs(Y3 - (m3 * X3 + c3))
print(f"Residuals: {residuals}")

# finding points with largest and smallest residuals

largest_residual_index = np.argmax(residuals)
largest_residual = np.max(residuals)
largest_residual_point = (X3[largest_residual_index], Y3[largest_residual_index])

smallest_residual_index = np.argmin(residuals)
smallest_residual = np.min(residuals)
smallest_residual_point = (X3[smallest_residual_index], Y3[smallest_residual_index])

print(f"Largest Residual: {largest_residual} at point {largest_residual_point}")
print(f"Smallest Residual: {smallest_residual} at point {smallest_residual_point}")


# plotting the data points and the fitted line

plt.figure()
plt.scatter(X3, Y3, color="blue", label="Data Points")
plt.plot(X3, m3 * X3 + c3, color="red", label="Fitted Line")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Best Fit Line")
plt.legend()
plt.show()
```
