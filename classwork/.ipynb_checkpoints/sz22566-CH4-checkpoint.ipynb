```python
import numpy as np
import math

i = []
counter = 99
leibniz = 0

# calculate the first 5000 terms in the Leibniz Series for pi

for n in range(0,5001):
    leibniz += ((-1)**n)/(2*n+1)  
    approx = 4 * leibniz
    i.append(approx)
    
    # print the relative error between Leibniz and np.pi every 100 terms 

    if n == counter:  
        print(((i[n] - np.pi)/np.pi)*100)
        counter += 100

```
