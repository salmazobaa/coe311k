```python
import numpy as np
import matplotlib.pyplot as plt

# Define the function and the domain
f = lambda x: np.sin(x)
a, b = 0, 2 * np.pi # Interval
N = 50 # Number of sample points


# Create a uniform grid
x = np.linspace(a, b, N)
y = f(x)

# Step size
h = (b - a) / (N - 1)

# Central Difference Approximation for the derivative
df_dx = (f(x + h) - f(x - h)) / (2 * h)

# Forward Difference Approximation
df_dx2 = (f(x + h) - f(x)) / (h)

# Backward Difference Approximation
df_dx3 = (f(x) - f(x - h)) / (h)


# Plot the approximation and the true derivative
fig = plt.figure(figsize=(12, 8))
fig.suptitle('Different Methods of Approximating the Derivative of sin(x)')

plt.subplot(2, 2, 1)
plt.plot( x , np.cos(x), label= 'Exact Derivative')
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.legend()

plt.subplot(2, 2, 2)
plt.plot(x, df_dx, 'o', label='Central Difference Approximation')
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.legend()

plt.subplot(2, 2, 3)
plt.plot(x, df_dx2, 'g*', label='Forward Difference Approximation')
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.legend()

plt.subplot(2, 2, 4)
plt.plot(x, df_dx3, 'rD', label='Backward Difference Approximation')
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.legend()

plt.show()
```
