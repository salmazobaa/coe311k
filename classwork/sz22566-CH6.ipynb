```python
import numpy as np
import time

def jacobi(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]

        # Check for convergence
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k + 1
        x = x_new
    return x, max_iterations

def gauss_seidel(A, b, x0, tol=1e-6, max_iterations=100):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = x.copy()
        for i in range(n):
            s1 = sum(A[i][j] * x_new[j] for j in range(i))       # updated values
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n))   # old values
            x_new[i] = (b[i] - s1 - s2) / A[i][i]

        # Check convergence
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k + 1
        x = x_new
    return x, max_iterations

A = np.array([[10, -1, 2, 0,0],
            [-1, 11, -1, 3, 0],
            [2, -1, 10, -1, 0],
            [0, 3, -1, 8, -2],
            [0, 0, 0, -2, 9]])

b = np.array([14, 30, 26, 25, 37])
x0 = np.zeros_like(b)
tol = 1e-6
max_iterations = 10000

start = time.time()
solution1, iterations1 = gauss_seidel(A, b, x0, tol, max_iterations)
end = time.time()
gauss_time = end - start

start = time.time()
solution2, iterations2 = jacobi(A, b, x0, tol, max_iterations)
end = time.time()
jacobi_time = end - start

print("gauss\' method")
print(f"Solution: {solution1}")
print(f"Iterations: {iterations1}")
print(f"time: {gauss_time:.6} seconds")

print()
print("Jacobi's")
print(f"Solution: {solution2}")
print(f"Iterations: {iterations2}")
print(f"time: {jacobi_time:.6} seconds")


A1 = np.array([[1, 2, 3, 0,0],
            [2, 1, 2, 3, 0],
            [3, 2, 1, 2, 0],
            [0, 3, 2, 1, 2],
            [0, 0, 3, 2, 1]])

b1 = np.array([14, 22, 33, 26, 22])

start = time.time()
solution3, iterations3 = gauss_seidel(A1, b1, x0, tol, max_iterations)
end = time.time()
gauss_time1 = end - start

start = time.time()
solution4, iterations4 = jacobi(A1, b1, x0, tol, max_iterations)
end = time.time()
jacobi_time1 = end - start

print("gauss\' method")
print(f"Solution: {solution3}")
print(f"Iterations: {iterations3}")
print(f"time: {gauss_time1:.6} seconds")

print()
print("Jacobi's")
print(f"Solution: {solution4}")
print(f"Iterations: {iterations4}")
print(f"time: {jacobi_time1:.6} seconds")


# second Ax=b does not converge because it is not diagonally dominant
# and uses all iterations available always.
```
