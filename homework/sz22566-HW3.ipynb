```python
import numpy as np
import math
import matplotlib.pyplot as plt


def taylor_series(N, x):
    total = 0
    for n in range(0, N):
        numerator = ((-1)**n)*(x**(2*n+1))
        denominator = math.factorial(2*n+1)
        result = numerator/denominator
        total += result
    return float(total)


x_values = []
errors_1_term = []
errors_2_term = []
errors_3_term = []
errors_4_term = []


for x in np.arange(0,10.1,0.1):
    x_values.append(x)
    for k in range(1,5):
        approx = taylor_series(k, x)
        abs_err = abs(np.sin(x) - approx)
        
        if k == 1:
            errors_1_term.append(abs_err)
        elif k == 2:
            errors_2_term.append(abs_err)
        elif k == 3:
            errors_3_term.append(abs_err)
        elif k == 4:
            errors_4_term.append(abs_err)

plt.plot(x_values, errors_1_term, label="1 term")
plt.plot(x_values, errors_2_term, label="2 terms")
plt.plot(x_values, errors_3_term, label="3 terms")
plt.plot(x_values, errors_4_term, label="4 terms")

plt.xlabel("x values")
plt.ylabel("Absolute Error")
plt.title("Truncation Error of Taylor Series Approximation of sin(x)")
plt.legend()
plt.show()



# The truncation error decreases as more terms are included. 
# The error decreases uniformly as x approaches zero.
# The truncation error is much less for small values of x
# versus large values of x.
```
